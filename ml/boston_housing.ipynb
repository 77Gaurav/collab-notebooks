```python
# Dataset:
# - 489 samples and 4 columns.
# - 3 input features: RM, LSTAT and PTRATIO.
# - MEDV is the target variable.
# - RM = average number of rooms per dwelling.
# - LSTAT = percentage of lower-status population.
# - PTRATIO = pupil-teacher ratio.
# - MEDV = median value of owner-occupied homes in $1000s.
```

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

file_path = "housing.csv"

df = kagglehub.load_dataset(KaggleDatasetAdapter.PANDAS,
                            "schirmerchad/bostonhoustingmlnd",
                            file_path)
```

```python
print(list(df.columns))
"""
RM = ROOMS
LSTAT = LOWER ECONOMY PEOPLE %
PTRATIO = STUDENTS TO TEACHERS RATIO
MEDV = MEDIAN VALUE OF HOUSE
"""
```

```python
X = df.drop("MEDV",axis=1)
y = df["MEDV"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test  = train_test_split(X,y,random_state=42,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression

model  = LinearRegression()
model.fit(X_train,y_train)

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error , r2_score

import numpy as np

print("MAE : ", mean_absolute_error(y_test,y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R²:", r2_score(y_test, y_pred))
```
